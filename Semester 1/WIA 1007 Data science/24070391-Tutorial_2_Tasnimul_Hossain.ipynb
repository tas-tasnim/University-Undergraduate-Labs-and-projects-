```python
arr = np.array([1,4,5,67,3], ndmin=5)
```

```python
def do_rest(arr):
    print('The array is : ')
    print(arr)
    print('the shape is = ',arr.shape)
    print('The dimension is = ', arr.ndim)
    print('The data type is = ', arr.dtype)
```

Question 1 : Create a numpy array of size 2 x 2

```python
arr_1 = np.array([[2,5],[5,5]])

print('a numpy array of size 2 x 2 --->>')
print(arr_1)
print('the dimension : ', arr_1.ndim)
```

Question 2 : Create a random numpy array of size 3 x 4


```python
arr_2 = np.random.rand(3,4)

do_rest(arr_2)
```

Question 3. Create a numpy array of ones and a numpy array of zeroes


```python
arr_0 = np.zeros((3,4))
arr_1 = np.ones((3,4))

do_rest(arr_0)
do_rest(arr_1)
```

Question 4. Create a random numpy array of size 6x3 and transpose it


```python
arr_4 = np.random.rand(6,3)

do_rest(arr_4)
trasposed_arr = arr_4.T
print('The trasposed array is : ')
do_rest(trasposed_arr)
```

Question 5 : Create two random numpy array of size n x p and p x m and then compute the dot product


```python
m= 5
n = 4
p = 6

arr_5 = np.random.rand(n,p)
arr_6 = np.random.rand(p,m)

print('The first array is : ')
do_rest(arr_5)
print('The second array is : ')
do_rest(arr_6)

product = np.dot(arr_5,arr_6)
print('The dot product is : ')
do_rest(product)
```
